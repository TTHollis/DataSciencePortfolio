```python
# Importing Libraries
import os
import requests
from io import BytesIO
from getpass import getpass
from PIL import Image
from IPython.display import display, Markdown, HTML

display(HTML('''
<style>
    div.output_subarea { page-break-inside: avoid; }
    div.jp-MarkdownOutput { page-break-inside: avoid; }
    div.cell { page-break-inside: avoid; }
</style>
'''))

# API key
key_path = os.path.expanduser('~/.stability/api_key.txt')
if os.path.exists(key_path):
    with open(key_path, 'r') as key_file:
        STABILITY_KEY = key_file.read().strip()
else:
    STABILITY_KEY = getpass('Enter your Stability AI API key: ')

# Stability AI REST v2beta endpoints
GENERATE_HOST = 'https://api.stability.ai/v2beta/stable-image/generate/sd3'
EDIT_HOST = 'https://api.stability.ai/v2beta/stable-image/edit/search-and-replace'

# Output folder for images saved during this notebook
output_dir = 'api_images'
os.makedirs(output_dir, exist_ok=True)

# Reusable request helper for the Stability REST API
def send_request(host, files, data):
    response = requests.post(
        host,
        headers={'Authorization': f'Bearer {STABILITY_KEY}', 'Accept': 'image/*'},
        files=files,
        data=data,
    )
    if not response.ok:
        raise Exception(f'Request failed ({response.status_code}): {response.text}')
    return response

print('Setup complete. Stability AI endpoints ready.')
print(f'Generate endpoint: {GENERATE_HOST}')
print(f'Edit endpoint:     {EDIT_HOST}')
print(f'Saving images to:  {output_dir}/')
```

### **Overview**

This section uses the Stability AI REST API (v2beta) to generate an image from a text prompt and then edit that generated image programmatically, all from within this notebook. The workflow has two API calls:

1. **Generate** - send a text prompt to the `stable-image/generate/sd3` endpoint using the `sd3.5-large` model to produce a new image.
2. **Edit** - send the generated image to the `stable-image/edit/search-and-replace` endpoint, naming an element to find and an element to swap in, to produce a modified version.

Holding the model constant at `sd3.5-large` keeps this consistent with the Stable Diffusion images created through the DreamStudio interface in the first part of the assignment, so the only thing changing is the method of access (point-and-click interface versus direct API calls). A closing narrative reflects on the results and on how straightforward the API was to work with.

### **Step 1: Generate an Image**

The first call sends a text prompt to the `generate/sd3` endpoint with the model set to `sd3.5-large`. The Stability API expects multipart form data rather than JSON, so the prompt and parameters are passed as form fields. A successful response returns the raw image bytes, which are saved to the output folder and displayed below.

```python
# Step 1: generate an image from a text prompt
prompt = ('a single red apple on a wooden table, plain background, '
          'soft natural lighting, photorealistic, centered')

response = send_request(
    GENERATE_HOST,
    files={'none': ''},
    data={
        'prompt': prompt,
        'model': 'sd3.5-large',
        'aspect_ratio': '2:3',
        'output_format': 'png',
    },
)

# Save the returned image 
generated_path = os.path.join(output_dir, 'api_generated.png')
with open(generated_path, 'wb') as image_file:
    image_file.write(response.content)

print(f'Image generated and saved to: {generated_path}')
print(f'Credits remaining (per response header): {response.headers.get("x-credits-remaining", "not reported")}')

# Display the generated image
display(HTML(f'<div style="width:50%; overflow:hidden;"><img src="{generated_path}" style="width:100%; margin-top:-20%;"></div>'))
```

### **Step 2: Edit the Generated Image**

The second call sends the image just generated to the `edit/search-and-replace` endpoint. This edit type takes the source image, a `search_prompt` naming the element to locate, and a `prompt` describing what to put in its place. The model finds the named element and replaces it, returning a modified image. This demonstrates editing an existing image through the API rather than generating from scratch.

```python
# Step 2: edit the generated image via search-and-replace
search_prompt = 'red apple'
replace_prompt = 'green pear'

with open(generated_path, 'rb') as source_image:
    response = send_request(
        EDIT_HOST,
        files={'image': source_image},
        data={
            'search_prompt': search_prompt,
            'prompt': replace_prompt,
            'output_format': 'png',
        },
    )

# Save the edited image
edited_path = os.path.join(output_dir, 'api_edited.png')
with open(edited_path, 'wb') as image_file:
    image_file.write(response.content)

print(f'Edited image saved to: {edited_path}')
print(f'Searched for: "{search_prompt}"  ->  replaced with: "{replace_prompt}"')
print(f'Credits remaining (per response header): {response.headers.get("x-credits-remaining", "not reported")}')

# Display the edited image
display(HTML(f'<div style="width:50%; overflow:hidden;"><img src="{edited_path}" style="width:100%; margin-top:-20%;"></div>'))
```
