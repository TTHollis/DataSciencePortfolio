### **Activity 3.01 — Generating Statistics from the Boston Housing Dataset**

In this activity, I will work with the `Boston_housing.csv` file to perform basic data‑wrangling and exploratory analysis. The goal is to load the dataset, inspect its structure, create a reduced DataFrame, visualize trends, and compute several summary statistics.

---

### **Steps to complete this activity:**

1. Load the Boston Housing dataset into a pandas DataFrame.  
2. Display the first 10 records and determine the total number of rows.  
3. Create a smaller DataFrame by removing the columns **CHAS**, **NOX**, **B**, and **LSTAT**.  
4. Display the last seven records of the reduced DataFrame.  
5. Plot histograms for all variables in the reduced DataFrame.  
6. Re‑create the histograms using a loop, adding unique titles.  
7. Create a scatter plot of **CRIM vs PRICE**.

<hr style='height:2px; background-color:#444; border:none; margin-top:30px; margin-bottom:25px;'>

```python
# Activity 3.01: Working with the Boston Housing Dataset

# 1. Display the first 10 records
print("First 10 Records:")
display(boston.head(10))

# 2. Total number of records
print("\nTotal Number of Records:", len(boston))

# 3. Create a smaller DataFrame without CHAS, NOX, B, and LSTAT
df_small = boston.drop(columns=['CHAS', 'NOX', 'B', 'LSTAT'])

# 4. Display the last 7 records of the reduced DataFrame
print("\nLast 7 Records of Reduced DataFrame:")
display(df_small.tail(7))

# 5. Plot histograms of all variables in the reduced DataFrame
df_small.hist(figsize=(12, 10), bins=20)
plt.suptitle("Histograms of Reduced Boston Housing Data", y=1.02)
plt.tight_layout()
plt.show()

# 6. Plot histograms using a loop with unique titles
for col in df_small.columns:
    plt.figure()
    plt.hist(df_small[col], bins=20)
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# 7. Scatter plot: Crime Rate vs Price
plt.scatter(boston['CRIM'], boston['PRICE'])
plt.xlabel('Crime Rate (CRIM)')
plt.ylabel('Price (PRICE)')
plt.title('Crime Rate vs Price')
plt.show()

# 8. Scatter plot: log10(CRIM) vs Price
plt.scatter(np.log10(boston['CRIM']), boston['PRICE'])
plt.xlabel('log10(CRIM)')
plt.ylabel('Price')
plt.title('log10(CRIM) vs Price')
plt.show()

# 9. Calculate summary statistics
mean_rooms = boston['RM'].mean()
median_age = boston['AGE'].median()
mean_distance = boston['DIS'].mean()
pct_low_price = (boston['PRICE'] < 20).mean() * 100

print("\nSummary Statistics:")
print(f"Mean rooms per dwelling: {mean_rooms:.2f}")
print(f"Median age: {median_age:.2f}")
print(f"Mean distance to employment centers: {mean_distance:.2f}")
print(f"Percentage of houses priced below $20,000: {pct_low_price:.2f}%")
```

### **Activity 4.01 — Working with the Adult Income Dataset (UCI)**

In this activity, I will work with the `adult_income_data.csv` file to explore, clean, and analyze the Adult Income dataset. The goal is to practice subsetting, applying user‑defined functions, detecting outliers, generating summary statistics, and merging DataFrames.

---

### **Steps to complete this activity:**

1. Load the Adult Income dataset into a pandas DataFrame.  
2. Read a text file line‑by‑line to demonstrate basic file handling.  
3. Add a response variable name (**Income**) to the dataset.  
4. Identify missing values.  
5. Create a subset containing **age**, **education**, and **occupation**.  
6. Plot a histogram of age with a bin size of 20.  
7. Create a function to strip whitespace and apply it to all string columns.  
8. Count the number of people aged between 30 and 50.  
9. Group by age and education to examine mean age distribution.  
10. Group by occupation to compute summary statistics of age, including:  
    - Which occupation has the oldest workers on average  
    - Which occupation has the largest share of workers above the 75th percentile  
11. Use subsetting and groupby to detect outliers.  
12. Plot the outlier values in a bar chart.  
13. Merge two DataFrames using common keys to remove duplicates.

<hr style='height:2px; background-color:#444; border:none; margin-top:30px; margin-bottom:25px;'>

```python
# 1. Load dataset with correct column names
adult = pd.read_csv(
    adult_path,
    header=None,
    names=[
        'age', 'workclass', 'fnlwgt', 'education', 'education_num',
        'marital_status', 'occupation', 'relationship', 'sex',
        'capital_gain', 'capital_loss', 'hours_per_week',
        'native_country', 'Income'
    ]
)

# 2. Display the first few rows and dataset shape
print("Initial DataFrame:\n")
display(adult.head())
print("\nDataset shape:", adult.shape)

# 3. Read a text file line-by-line (demonstration)
print("\nReading first 5 lines of the dataset file:")
with open(adult_path, 'r') as f:
    for i, line in enumerate(f):
        if i < 5:
            print(line.strip())

# 4. Identify missing values
print("\nMissing Values:\n")
print(adult.isna().sum())

# 5. Create a subset with age, education, occupation
adult_sub = adult[['age', 'education', 'occupation']]

print("\nSubset (age, education, occupation):\n")
display(adult_sub.head())

# 6. Plot histogram of age with bin size 20
plt.figure(figsize=(8, 5))
plt.hist(adult['age'], bins=20, color='steelblue')
plt.title('Age Distribution (bins=20)')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# 7. Function to strip whitespace


def strip_whitespace(x):
    if isinstance(x, str):
        return x.strip()
    return x


# Apply to all string columns
for col in adult.columns:
    if adult[col].dtype == 'object':
        adult[col] = adult[col].apply(strip_whitespace)

# 8. Count people aged 30–50
count_30_50 = adult[(adult['age'] >= 30) & (adult['age'] <= 50)].shape[0]
print("\nNumber of people aged 30–50:", count_30_50)

# 9. Group by age and education to examine mean age distribution
print("\nMean Age by Age and Education:\n")
display(adult.groupby(['age', 'education'])['age'].mean())

# 10. Group by occupation: summary statistics of age
summary = adult.groupby('occupation')['age'].describe()
print("\nSummary Statistics of Age by Occupation:\n")
display(summary)

# Occupation with oldest workers on average
oldest_avg = summary['mean'].idxmax()
print("\nOccupation with oldest workers on average:", oldest_avg)

# Occupation with largest share above 75th percentile
p75 = adult['age'].quantile(0.75)

share_above_75 = adult.groupby('occupation')['age'].apply(
    lambda s: (s > p75).mean()
)

largest_share = share_above_75.idxmax()

print("Occupation with largest share above 75th percentile:", largest_share)

# 11. Detect outliers (age > 75th percentile)
outliers = adult[adult['age'] > p75]
outliers_by_occ = outliers.groupby('occupation')['age'].count()

print("\nOutliers by Occupation:\n")
display(outliers_by_occ)

# 12. Bar plot of outliers
plt.figure(figsize=(10, 5))
outliers_by_occ.plot(kind='bar', color='darkred')
plt.title('Outliers by Occupation (Age > 75th Percentile)')
plt.xlabel('Occupation')
plt.ylabel('Count of Outliers')
plt.tight_layout()
plt.show()

# 13. Merge two DataFrames using common keys
merged = pd.merge(adult, adult_sub, on=['age', 'education', 'occupation'])

print("\nMerged DataFrame:\n")
display(merged.head())
```

### **Summary**

This activity demonstrated how to load and explore the Adult Income dataset, inspect its structure, and perform a series of data‑wrangling tasks. After assigning proper column names, I examined missing values, created a focused subset of key variables, and visualized the age distribution. I also applied a custom whitespace‑stripping function, analyzed demographic patterns through grouping operations, identified outliers based on age, and displayed these outliers using a bar chart. Finally, I merged DataFrames using common keys to reinforce dataset consistency and practice joining operations.

### **Q3: Creating Series and Index Labels and Performing Arithmetic Operations**

For this activity, I will create two pandas series with different index labels and perform arithmetic operations that demonstrate pandas’ automatic index alignment.

### **Steps to complete**

1. Create Series 1 with values 7.3, -2.5, 3.4, 1.5, and indices a, c, d, e.  
2. Create Series 2 with values -2.1, 3.6, -1.5, 4, 3.1, and indices a, c, e, f, g.  
3. Add the two Series together.  
4. Subtract Series 1 from Series 2.  
5. Print all results.
6. Create a table to highlight `NaN` values

```python
# 1. Create Series 1
series1 = pd.Series([7.3, -2.5, 3.4, 1.5], index=['a', 'c', 'd', 'e'])

# 2. Create Series 2
series2 = pd.Series([-2.1, 3.6, -1.5, 4, 3.1], index=['a', 'c', 'e', 'f', 'g'])

# 3. Add the two Series
addition_result = series1 + series2

# 4. Subtract Series 1 from Series 2
subtraction_result = series2 - series1

# 5. Print results
print("Series 1:\n", series1)
print("\nSeries 2:\n", series2)
print("\nAddition Result:\n", addition_result)
print("\nSubtraction Result:\n", subtraction_result, "\n")

# 6. Create combined table view
combined = pd.DataFrame({
    'Series 1': series1,
    'Series 2': series2,
    'Addition': addition_result,
    'Subtraction': subtraction_result
})

styled = (
    combined.style
    .format('{:.1f}')
    .highlight_null(color='red')
    .set_caption('Series Arithmetic Table')
    .set_table_styles([
        {'selector': 'caption', 'props': [
                ('font-size', '16px'),
                ('font-weight', 'bold')
        ]},
        {'selector': 'th', 'props': [
            ('font-weight', 'bold'),
            ('text-align', 'center')
        ]}
    ])
)

styled
```

### **Summary**

In this section, I created two pandas Series with mismatched indices:

- **Series 1**: values = [7.3, -2.5, 3.4, 1.5], indices = ['a', 'c', 'd', 'e']
- **Series 2**: values = [-2.1, 3.6, -1.5, 4, 3.1], indices = ['a', 'c', 'e', 'f', 'g']

I then performed basic arithmetic operations:

- **Addition** (`Series 1 + Series 2`)
- **Subtraction** (`Series 2 - Series 1`)

**Key observations**:
- Pandas automatically aligns the operations based on index labels rather than position.
- Where indices match (a, c, e), the arithmetic produces expected numeric results.
- Where an index exists in only one Series (d in Series 1; f and g in Series 2), the result is **NaN** which basically is missing data; this allows combining datasets with different (or incomplete) keys without manual merging or looping.

The visualization (side-by-side table with highlighted NaNs) clearly demonstrates this alignment rule and makes it easy to spot where data is missing.
