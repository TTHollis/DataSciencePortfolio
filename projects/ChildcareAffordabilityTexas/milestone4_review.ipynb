## The Childcare Affordability Crisis in Texas

<br>

#### Summary of Analysis

The national database of childcare prices (U.S. Department of Labor, Women's Bureau, 2024) contains 34,567 county-year records spanning 2008 through 2018 across 51 jurisdictions, with 227 columns. Working in Python, I narrowed the file to identifiers, median weekly center-based infant and preschool prices, median household income, and female median earnings, then annualized the prices and measured them against the U.S. Department of Health and Human Services standard that childcare is affordable at no more than 7% of household income.

#### Findings

Three findings anchor the story. First, childcare fails the federal standard almost universally: in 2018 only 4 of 2,360 reporting counties nationwide came in at or below 7%, the median county spent 16.1% of household income on one infant's care, and zero of 254 Texas counties met the standard. Second, the burden grew across the decade, nationally from 15.4% (2008) to 16.1% (2018) and in Texas from 13.3% (2009) to 14.5% (2018), even as the economy recovered. Third, the household math is severe: one infant consumes 33.2% of female median earnings in the typical county and 31.7% in Texas, while an infant plus a preschooler consumes 61.1% nationally and 58.3% in Texas. Bexar County sits at $8,666 per year for infant care, or 33.1% of female median earnings. A test of care burden against labor force participation among mothers of young children found essentially no relationship (r = -0.04), which I report rather than omit.

#### Assumptions

I assumed that median center-based prices represent the typical family's cost, that care is purchased full time for 52 weeks per year, and that nominal prices should be compared against nominal income within the same year. Female median earnings include part-time workers, which raises the earnings-share figures above what they would be against full-time salaries.

#### Items That Still Need Clarification

Roughly a third of price values are missing, and reporting coverage grew from 1,424 counties in 2008 to 2,360 in 2018, so trend statements must be read alongside changing coverage. Texas reporting does not begin until 2009. The dataset flags imputed values, and I have not yet quantified how much of each visual rests on modeled rather than observed prices. Establishing whether high prices push mothers out of the workforce would require within-county longitudinal analysis, and the data ends in 2018, so it predates the pandemic.

<br><br>

#### Direction of Story and Target Audience

The story is persuasive rather than informational: childcare in Texas fails the federal government's own affordability standard everywhere it is measured, the burden is growing, and the math now forces families to weigh whether a second income still pays. The ask is specific, to expand state childcare subsidy funding and raise the income-eligibility ceiling. The audience is Texas state legislators and their staff, policy literate but not data literate and fiscally skeptical toward new spending. My role is a constituent data analyst presenting verified federal data, and the story counters that skepticism by framing childcare as workforce infrastructure rather than welfare.

#### Mediums Included and Why

Three mediums carry the story at three speeds. A PowerPoint briefing deck walks a legislator through the standard, the gap, the trend, and the second-earner math before the ask appears. A Power BI dashboard lets staff skip the narration and look up their own district. A one-page infographic distills it into a leave-behind that outlives the meeting.

#### Design Decisions

All three mediums share one colorblind-safe palette, with amber reserved exclusively for the 7% benchmark. Slide titles are written as complete claims rather than topic labels, so a slide forwarded in isolation still argues something. Axes are zero-based so the gap above the benchmark is not exaggerated, and scope is assigned by medium: the deck builds the national case before localizing, while the dashboard and infographic carry Texas values throughout.

The most significant item carried into the final deliverable is the Milestone 3 feedback that the same charts appeared across mediums. The direction I am working toward is a distinct set of visuals per medium rather than resized versions of shared ones, with the deck holding the national distribution and trend views and the dashboard and infographic moving toward county-level and non-chart treatments of the Texas data. Specific chart types are still being evaluated, but the governing rule is that no visual repeats across mediums and each draws on its own cut of the data.

#### Ethical Considerations

I added no imputation of my own. I dropped county-years missing infant price or income, which removed about 32% of rows, and that count and the coverage growth are disclosed alongside any trend claim. Derived fields are limited to annualized cost and income share, with no outlier removal or smoothing. The source is a public federal dataset carrying no personally identifiable information and no licensing restriction, acquired directly from the publishing agency and cited in every artifact. I verified counts against the published documentation and sanity-checked medians against known national estimates.

The real risks live in the transformations. Assuming 52 weeks overstates cost for part-year care, comparing full-time care against earnings that include part-time workers inflates the share, and dropping non-reporting counties biases the sample toward larger, more urban places. Persuasive framing could also imply that prices cause mothers to leave the workforce, which my own correlation does not support, and pricing care against a mother's earnings encodes an assumption about who the second earner is. My intent is to mitigate these by footnoting assumptions where they appear, reporting the null correlation openly, keeping axes zero-based, and offering full methodology on request.

#### Lessons Learned

The clearest lesson is that medium-specific visuals have to be planned at the storyboard stage. Because my Milestone 2 mock-ups reused the same chart shapes, the drafts inherited that duplication. Next time I will assign each medium a distinct data cut before drawing anything, and quantify imputation coverage earlier, since it affects how strongly I can phrase a trend claim.

What I enjoyed most was the moment the story stopped being a national statistic and became a local one. Watching the abstract 7% standard turn into $8,666 a year in Bexar County made the case for keeping one concrete, local number at the center of any persuasive data story.
